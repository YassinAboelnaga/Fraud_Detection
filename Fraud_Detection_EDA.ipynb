```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8,5))
x=df[df['Amount']<1000]
plt.hist(x['Amount'], bins=30)

plt.xlabel("Transaction Amount")
plt.ylabel("Number of Transactions")
plt.title("Distribution of Transaction Amount")

plt.show()
```
